```python
import pandas as pd

# Read CSV file
text_df = pd.read_csv("googleplaystore.csv")

# Read Excel file
excel_df = pd.read_excel("file_example_XLS_10.xls")

# Read Web data
web_df = pd.read_csv(
    "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
)

# Display first 5 rows
print("CSV DATA")
print(text_df.head())

print("\nEXCEL DATA")
print(excel_df.head())

print("\nWEB DATA")
print(web_df.head())

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# Save processed files
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nProcessed files saved successfully!")
```
