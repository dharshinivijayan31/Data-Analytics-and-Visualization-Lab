```python
import pandas as pd

df = pd.read_csv("data.csv")

print("First 5 rows:\n", df.head())
print("Last 5 rows:\n", df.tail())

df.info()

print("Summary statistics:\n", df.describe())

# Handle missing values
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())

# Create new column
df['new_column'] = df['Rating'] * 2

# Series operation
series = df['Rating']
print("Series addition:\n", series + 10)

# Filter rows
filtered_df = df[df['Rating'] > 4.0]
print("Filtered DataFrame:\n", filtered_df)

# Grouping
grouped = df.groupby('Category')['Rating'].mean()
print("Grouped mean:\n", grouped)

# Sorting
df_sorted = df.sort_values(by='Rating', ascending=False)
print("Sorted DataFrame:\n", df_sorted)

# Boolean masking
masked_df = df[df['Rating'] > df['Rating'].median()]
print("Masked DataFrame:\n", masked_df)

# Remove duplicates and missing values
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# Subset DataFrame
subset_df = df[['App', 'Category']]

# Save to CSV
subset_df.to_csv('filtered_data.csv', index=False)

# Summary statistics
print("Total sum:", df['Rating'].sum())
print("Mean:", df['Rating'].mean())
print("Standard Deviation:", df['Rating'].std())
```
